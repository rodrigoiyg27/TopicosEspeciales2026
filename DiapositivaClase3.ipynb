```python
import pandas as pd

df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")

# --- .iloc: posición entera (como listas de Python) ---
primera_fila = df.iloc[0]                  # la fila en la posición 0
primeras_5 = df.iloc[0:5]                  # posiciones 0,1,2,3,4 (5 EXCLUIDO)
celda = df.iloc[0, 2]                      # fila 0, columna en posición 2

# --- .loc: etiqueta ---
# Como el índice por defecto es un RangeIndex (0,1,2...) numérico,
# aquí coincide visualmente con .iloc, pero conceptualmente es distinto.
fila_por_label = df.loc[0]                 # fila CON ETIQUETA 0
rango_por_label = df.loc[0:5]              # etiquetas 0 a 5 (5 SÍ INCLUIDO) → 6 filas

# .loc también permite elegir columnas por nombre, filas y columnas a la vez
subset = df.loc[0:5, ["Invoice", "Description", "Quantity"]]

print(subset)
```

```python
#Filtrado
import pandas as pd
df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")
# Paso 1: la condición genera una Series booleana, una por fila
condicion = df["Quantity"] > 0
print(condicion.head())   # True / False por cada fila

# Paso 2: esa Series booleana se usa como "máscara" para filtrar
ventas_positivas = df.loc[condicion]

# En la práctica, casi siempre se escribe todo en una sola línea:
ventas_positivas = df.loc[df["Quantity"] > 0]

# --- Condiciones combinadas: se usa & (y) y | (o), NUNCA 'and'/'or' de Python ---
# IMPORTANTE: cada condición individual debe ir entre paréntesis
ventas_grandes_uk = df.loc[
    (df["Quantity"] > 50) & (df["Country"] == "United Kingdom")
]

# --- Filtrar por texto: útil para explorar productos por nombre ---
# .str.contains() busca la subcadena dentro de cada valor de texto
regalos = df.loc[
    df["Description"].str.contains("GIFT", case=False, na=False)
    # na=False: evita que las filas con Description vacía (NaN) rompan el filtro
]

print("Filas con cantidad negativa (posibles devoluciones):", (df["Quantity"] < 0).sum())
```

```python
import pandas as pd

df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")

# --- Ordenar por una sola columna, de mayor a menor ---
por_cantidad = df.sort_values(by="Quantity", ascending=False)

# --- Ordenar por varias columnas a la vez ---
# Primero por país (alfabético), y dentro de cada país por cantidad descendente
por_pais_y_cantidad = df.sort_values(
    by=["Country", "Quantity"],
    ascending=[True, False]   # una dirección por cada columna en 'by'
)

# --- Top 10 filas con mayor cantidad vendida  ---
top_10 = df.sort_values(by="Quantity", ascending=False).head(10)

print(top_10[["Description", "Quantity", "Country"]])
```

```python
# Ordenamos por cantidad, de mayor a menor
df_ordenado = df.sort_values(by="Quantity", ascending=False)

# .iloc[0] siempre trae la PRIMERA fila FÍSICA del resultado ordenado
# → la de mayor Quantity, que es justamente lo que buscábamos al ordenar
print("iloc[0]:")
print(df_ordenado.iloc[0])

# .loc[0] busca la fila cuya ETIQUETA es 0 — es decir, la fila que ERA
# la primera ANTES de ordenar. Después de sort_values(), esa etiqueta
# puede estar en cualquier posición del nuevo orden.
print("\nloc[0]:")
print(df_ordenado.loc[0])
```

```python
import pandas as pd

df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")

# 1. Revisar los tipos actuales
print(df.dtypes)
# InvoiceDate suele venir como datetime64 (bien, porque Excel lo tipa),
# pero si el mismo dataset se carga desde CSV, InvoiceDate llega como texto (object).

# 2. Forzar la conversión correcta de una columna de fecha (caso CSV)
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

# 3. CustomerID llega como float aunque
#    conceptualmente es un identificador, no una cantidad continua.
#    Se deja como float por ahora: convertir a int fallaría mientras existan NaN
print(df["Customer ID"].dtype)

# 4. Convertir una columna de texto a categoría: reduce memoria cuando hay
#    pocos valores únicos repetidos muchas veces (caso típico de "Country")
print("Países únicos:", df["Country"].nunique())
df["Country"] = df["Country"].astype("category")

print(df.dtypes)
```
