```python
import pandas as pd

url_ventas = "https://raw.githubusercontent.com/rodrigoiyg27/TopicosEspeciales2026/main/ventas.csv"
url_clientes = "https://raw.githubusercontent.com/rodrigoiyg27/TopicosEspeciales2026/main/clientes.csv"

ventas = pd.read_csv(url_ventas)
clientes = pd.read_csv(url_clientes)

print("Ventas cargado:", ventas.shape)
print("Clientes cargado:", clientes.shape)
```

Exploración inicial

```python
# .head() muestra las primeras filas para confirmar que la carga fue correcta
print("Primeras filas de ventas")
display(ventas.head())

print("\nPrimeras filas de clientes ")
display(clientes.head())

# .info() muestra tipos de dato por columna y cuántos valores no nulos tiene cada una
print("\n Info de ventas ")
ventas.info()

print("\nInfo de clientes")
clientes.info()
```

Estadistica descriptivas generales

```python
# .describe() resume las columnas numéricas: promedio, mínimo, máximo, cuartiles
print("=== Estadísticas descriptivas de ventas ===")
display(ventas.describe())

# include='object' hace lo mismo pero para columnas de texto/categóricas:
# muestra cuántos valores únicos hay y cuál es el más frecuente
print("\n=== Estadísticas descriptivas de columnas categóricas ===")
display(ventas.describe(include='object'))
```

Comprobación de valores nulos

```python

print("Valores nulos en ventas ")
print(ventas.isnull().sum())

print("\n Valores nulos en clientes ")
print(clientes.isnull().sum())

#porcentaje de nulos por columna (útil cuando el dataset es más grande)
print("\n % de nulos por columna en ventas ")
print((ventas.isnull().mean() * 100).round(2))
```

Comprobación de duplicados

```python
# duplicated() marca como True cualquier fila que sea idéntica a una anterior
print("Filas totalmente duplicadas en ventas:", ventas.duplicated().sum())
print("Filas totalmente duplicadas en clientes:", clientes.duplicated().sum())

# También conviene revisar duplicados por una clave de negocio específica,
# Por ejemplo: ¿algún CustomerID
# aparece más de una vez en la tabla de clientes? (no debería, es su clave única)
duplicados_customerid = clientes[clientes.duplicated(subset='CustomerID', keep=False)]
print("\nCustomerID duplicados en clientes:")
display(duplicados_customerid)
```

Revisión de valores categóricos

```python
print("Países en ventas:", ventas['Country'].unique())
print("Países en clientes:", clientes['Country'].unique())
print("Categorías en ventas:", ventas['Categoria'].unique())
print("Segmentos en clientes:", clientes['Segmento'].unique())
```

Preparar columna de ingresos

```python
# Antes de agrupar necesitamos una métrica numérica que agregar.
# IngresosBrutos = cantidad vendida x precio unitario, por cada línea de venta
ventas["IngresosBrutos"] = ventas["Quantity"] * ventas["UnitPrice"]

ventas[["InvoiceNo", "Description", "Country", "Categoria", "IngresosBrutos"]].head()
```

Dividir en grupos

```python
# groupby() por sí solo NO calcula nada todavía: solo organiza las filas en grupos.
grupos_por_pais = ventas.groupby("Country")

# Podemos inspeccionar cuántos grupos se formaron y cuáles son
print("Grupos formados:", list(grupos_por_pais.groups.keys()))

# Y podemos ver, por ejemplo, qué filas cayeron en el grupo "France"
grupos_por_pais.get_group("France")
```

Aplicar y combinar

```python
# Al encadenar una función de agregación, pandas aplica esa función a CADA grupo
# por separado (apply) y luego junta los resultados en una sola tabla (combine)
ingresos_por_pais = ventas.groupby("Country")["IngresosBrutos"].sum()

print(ingresos_por_pais)
```

Verificación

```python
# Esto es exactamente lo mismo que la celda anterior, pero hecho
# grupo por grupo, para que quede clarísimo que groupby no hace magia:
# solo automatiza este mismo proceso.
for pais, grupo in ventas.groupby("Country"):
    total = grupo["IngresosBrutos"].sum()
    print(f"{pais}: ${total:,.2f}  (a partir de {len(grupo)} filas)")
```

Uso de apply

```python
# La fase "apply" puede ser cualquier función de agregación: sum, mean, count, max...
print("Promedio de ingreso por línea de venta, por país:")
print(ventas.groupby("Country")["IngresosBrutos"].mean())

print("\nNúmero de líneas de venta por país:")
print(ventas.groupby("Country")["IngresosBrutos"].count())
```

Agrupar por más de una clave a la vez

```python
# También podemos "dividir" usando más de una columna, pasando una lista.
# Aquí agrupamos por País Y Categoría al mismo tiempo.
ingresos_pais_categoria = ventas.groupby(["Country", "Categoria"])["IngresosBrutos"].sum()

print(ingresos_pais_categoria)
```

Mantener las claves como columnas normales

```python
# Por defecto, las columnas de agrupación se convierten en índice del resultado.
# Con as_index=False, se mantienen como columnas normales, lo cual suele ser
# más cómodo para seguir trabajando con el resultado (por ejemplo, para graficarlo después)
resumen = ventas.groupby(["Country", "Categoria"], as_index=False)["IngresosBrutos"].sum()

resumen
```

Diferencia entre size y count

```python
# .size() cuenta el número de FILAS del grupo, sin importar si hay nulos
# .count() cuenta valores NO NULOS de una columna específica

print("size():", ventas.groupby("Country").size())
print("\ncount() sobre CustomerID:", ventas.groupby("Country")["CustomerID"].count())
```

Múltiples métricas con groupby

```python
# Si quisiéramos varias métricas con groupby "simple", tendríamos que repetir
# la agrupación una y otra vez — ineficiente y poco legible
ingresos = ventas.groupby("Country")["IngresosBrutos"].sum()
unidades = ventas.groupby("Country")["Quantity"].sum()
num_facturas = ventas.groupby("Country")["InvoiceNo"].nunique()

print(ingresos)
print(unidades)
print(num_facturas)
```

Múltiples métricas con .agg()

```python
# .agg() permite calcular varias métricas en una sola pasada por los datos.
# Sintaxis: nombre_columna_resultado=("columna_origen", "funcion")
resumen_pais = ventas.groupby("Country").agg(
    ingresos_totales=("IngresosBrutos", "sum"),
    unidades_vendidas=("Quantity", "sum"),
    num_facturas=("InvoiceNo", "nunique"),   # nunique = facturas DISTINTAS, no líneas
    num_lineas=("InvoiceNo", "count"),        # count = total de líneas de venta
)

resumen_pais
```

nunique vs count

```python
# Una misma factura puede tener varias líneas (varios productos comprados juntos).
# nunique() cuenta facturas ÚNICAS; count() cuenta cada LÍNEA por separado.

uk = ventas[ventas["Country"] == "United Kingdom"]

print("Facturas distintas (nunique):", uk["InvoiceNo"].nunique())
print("Líneas de venta totales (count):", uk["InvoiceNo"].count())
print("\nEsto pasa porque varias líneas comparten el mismo InvoiceNo:")
print(uk["InvoiceNo"].value_counts())
```

Varias funciones en una misma columna

```python
# También se puede pedir varias funciones sobre una sola columna.
# Aquí usamos la sintaxis de lista en vez de tuplas nombradas.
estadisticas_ingresosBrutos = ventas.groupby("Country")["IngresosBrutos"].agg(["sum", "mean", "min", "max"])

estadisticas_ingresosBrutos
```

Agregaciones múltiples con doble clave de agrupación (Categoría por País)

```python

resumen_pais_categoria = ventas.groupby(["Country", "Categoria"], as_index=False).agg(
    ingresos_totales=("IngresosBrutos", "sum"),
    unidades_vendidas=("Quantity", "sum"),
    num_lineas=("InvoiceNo", "count"),
)
#País y categoría con mayores ventas
resumen_pais_categoria.sort_values("ingresos_totales", ascending=False)
```

Tipos de merge

Tablas originales

```python
print("Tabla ventas:", ventas.shape)
print(ventas[["InvoiceNo", "CustomerID", "Country"]].drop_duplicates())

print("\nTabla clientes:", clientes.shape)
print(clientes)
```

Primer merge: unión por customerID

```python
# Unimos por CustomerID. OJO: ambas tablas tienen una columna "Country",
# así que pandas va a renombrarlas automáticamente con sufijos _x / _y
primer_intento = ventas.merge(clientes, on="CustomerID", how="inner")

primer_intento[["InvoiceNo", "CustomerID", "Country_x", "Country_y", "Segmento"]].head()
```

Corrección

```python
# Como Country ya existe en ventas y es la misma información, no necesitamos
# traerla de nuevo desde clientes. Seleccionamos solo lo que aporta valor nuevo.
clientes_merge = clientes[["CustomerID", "Segmento"]]

ventas_con_cliente = ventas.merge(clientes_merge, on="CustomerID", how="inner")

ventas_con_cliente.head()
```

Inner Join

```python
inner = ventas.merge(clientes_merge, on="CustomerID", how="inner")

print("Filas resultantes (inner):", inner.shape[0])
print("Filas originales en ventas:", ventas.shape[0])
# Como TODOS los CustomerID de ventas existen en clientes, inner no pierde ninguna fila.
# El único que se queda fuera es el 16000, porque nunca aparece en ventas.
```

left Join

```python
left = ventas.merge(clientes_merge, on="CustomerID", how="left")

print("Filas resultantes (left):", left.shape[0])
# Da lo mismo que inner en este caso, porque no hay ningún CustomerID en ventas
# que falte en clientes. Left solo se nota cuando el lado izquierdo tiene huérfanos.
```

Right Join

```python
right = ventas.merge(clientes_merge, on="CustomerID", how="right")

print("Filas resultantes (right):", right.shape[0])
print("\nAquí SÍ aparece el cliente 16000, con NaN en las columnas de ventas:")
right[right["CustomerID"] == 16000]
```

Outer Join

```python
outer = ventas.merge(clientes_merge, on="CustomerID", how="outer")

print("Filas resultantes (outer):", outer.shape[0])
# outer = inner + los huérfanos de la izquierda + los huérfanos de la derecha
# En este caso da igual que right, porque ventas no tiene huérfanos propios.
```

Comparacion entre los 4 tipos de merge

```python
comparacion = pd.DataFrame({
    "tipo_union": ["inner", "left", "right", "outer"],
    "filas_resultantes": [inner.shape[0], left.shape[0], right.shape[0], outer.shape[0]],
})

comparacion
```

Verificar filas sin coincidencias despues del merge

```python
# Sin importar qué tipo de unión uses, SIEMPRE conviene revisar cuántas filas
# quedaron con NaN en las columnas que vinieron de la tabla derecha.
# Esto detecta problemas de unión antes de generar un reporte.
nulos_tras_merge = right["Segmento"].isna().sum()
print(f"Filas sin Segmento asignado tras el right join: {nulos_tras_merge}")

right[right["Segmento"].isna()]
```

Pivot Table

```python
# Esto ya lo hicimos con groupby: País x Categoría, en formato "largo"
resumen_largo = ventas.groupby(["Country", "Categoria"])["IngresosBrutos"].sum()
print(resumen_largo)
```

```python
tabla_pais_categoria = pd.pivot_table(
    ventas,
    index="Country",       # esto va en las FILAS
    columns="Categoria",   # esto va en las COLUMNAS
    values="IngresosBrutos",      # esto es lo que se agrega
    aggfunc="sum",         # cómo se agrega (igual que en groupby)
)

tabla_pais_categoria
```

manejar espacios blancos de información

```python
# Si algún país no tiene ventas en alguna categoría, esa celda queda en NaN.
# fill_value reemplaza esos huecos por un valor explícito (normalmente 0)
tabla_pais_categoria = pd.pivot_table(
    ventas,
    index="Country",
    columns="Categoria",
    values="IngresosBrutos",
    aggfunc="sum",
    fill_value=0,   # sin esto, Francia mostraría NaN si no tuviera alguna categoría
)

tabla_pais_categoria
```

Agregar totales de fila y columna

```python
# margins=True agrega una fila y una columna de "Total", muy típico en reportes ejecutivos
tabla_con_totales = pd.pivot_table(
    ventas,
    index="Country",
    columns="Categoria",
    values="IngresosBrutos",
    aggfunc="sum",
    fill_value=0,
    margins=True,          # agrega fila/columna "All"
    margins_name="Total",  # le cambiamos el nombre a "Total" en vez de "All"
)

tabla_con_totales
```

Múltiples métricas

```python
# aggfunc también puede recibir una lista, igual que en agg()
tabla_multi_metrica = pd.pivot_table(
    ventas,
    index="Country",
    columns="Categoria",
    values="IngresosBrutos",
    aggfunc=["sum", "mean", "count"],
    fill_value=0,
)

tabla_multi_metrica
```

Invirtiendo filas y columnas

```python
# Categoría en filas, País en columnas: mismo resultado, otra orientación.
# Útil para decidir cuál se lee mejor según cuántas categorías/países tengas.
tabla_invertida = pd.pivot_table(
    ventas,
    index="Categoria",
    columns="Country",
    values="IngresosBrutos",
    aggfunc="sum",
    fill_value=0,
)

tabla_invertida
```

groupby + reshape

```python
paso1 = ventas.groupby(["Country", "Categoria"])["IngresosBrutos"].sum()

# ..unstack() convierte
# el segundo nivel del índice en columnas, que es justo lo que hace pivot_table.
paso2 = paso1.unstack(fill_value=0)

print(paso2)
print("\nSon iguales?", paso2.equals(tabla_pais_categoria))
```

KPI

KPI #1: Ticket promedio

```python
# El "ticket" es el total de una factura completa, no de una sola línea de venta.
# Por eso primero agrupamos por Ingreso antes de calcular el promedio.
ingresos_por_factura = ventas.groupby("InvoiceNo")["IngresosBrutos"].sum()

print(ingresos_por_factura)

ticket_promedio = ingresos_por_factura.mean()
print(f"\nTicket promedio: ${ticket_promedio:,.2f}")
```

Ticket promedio (bruto vs neto)

```python
ticket_promedio_bruto = ingresos_por_factura.mean()

# Para la versión "neta", excluimos las facturas canceladas (las que empiezan con "C")
ventas_sin_cancelaciones = ventas[~ventas["InvoiceNo"].astype(str).str.startswith("C")]
ingresos_por_factura_neto = ventas_sin_cancelaciones.groupby("InvoiceNo")["IngresosBrutos"].sum()
ticket_promedio_neto = ingresos_por_factura_neto.mean()

print(f"Ticket promedio (bruto, con cancelaciones): ${ticket_promedio_bruto:,.2f}")
print(f"Ticket promedio (neto, sin cancelaciones):  ${ticket_promedio_neto:,.2f}")
```

KPI #2: Top n productos por ingresos

```python

top_productos = (
    ventas_sin_cancelaciones
    .groupby("Description")["IngresosBrutos"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

top_productos
```

Top-N clientes por ingresos

```python
# El mismo patrón, pero agrupando por cliente en lugar de por producto
top_clientes = (
    ventas_sin_cancelaciones
    .groupby("CustomerID")["IngresosBrutos"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

top_clientes
```

KPI #3: Tasa de cancelación

```python
# Marcamos qué facturas son cancelaciones (prefijo "C" en InvoiceNo)
ventas["EsCancelacion"] = ventas["InvoiceNo"].astype(str).str.startswith("C")

# La tasa de cancelación se calcula sobre FACTURAS distintas, no sobre líneas de venta
facturas_totales = ventas["InvoiceNo"].nunique()
facturas_canceladas = ventas.loc[ventas["EsCancelacion"], "InvoiceNo"].nunique()

tasa_cancelacion = facturas_canceladas / facturas_totales

print(f"Facturas totales: {facturas_totales}")
print(f"Facturas canceladas: {facturas_canceladas}")
print(f"Tasa de cancelación: {tasa_cancelacion:.1%}")
```

Tasa de cancelación por país

```python
cancelacion_por_pais = (
    ventas.groupby("Country")
    .agg(
        facturas_totales=("InvoiceNo", "nunique"),
        facturas_canceladas=("EsCancelacion", "sum"),  # sum de booleanos = conteo de True
    )
)

cancelacion_por_pais["tasa_cancelacion"] = (
    cancelacion_por_pais["facturas_canceladas"] / cancelacion_por_pais["facturas_totales"]
)

cancelacion_por_pais
```

KPI #4: Frecuencia de compra por cliente

```python
# ¿Cuántas facturas distintas generó cada cliente?
frecuencia_por_cliente = (
    ventas_sin_cancelaciones
    .groupby("CustomerID")["InvoiceNo"]
    .nunique()
    .sort_values(ascending=False)
)

frecuencia_por_cliente
```

Resumen ejecutivo

```python
print("=== Resumen ejecutivo ===")
print(f"Ticket promedio (neto): ${ticket_promedio_neto:,.2f}")
print(f"Producto más vendido: {top_productos.index[0]} (${top_productos.iloc[0]:,.2f})")
print(f"Tasa de cancelación global: {tasa_cancelacion:.1%}")
print(f"Cliente más frecuente: {frecuencia_por_cliente.index[0]} ({frecuencia_por_cliente.iloc[0]} facturas)")
```
