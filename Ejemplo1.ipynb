```python
import pandas as pd

df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")

# 1. ¿Cuántas filas y columnas tiene?
print("Forma del dataset:", df.shape)

# 2. ¿Cómo se llaman las columnas?
print("Columnas:", df.columns.tolist())

# 3. ¿Qué tipo de dato tiene cada columna? (número, texto, fecha...)
print(df.dtypes)

# 4. Primeras y últimas filas, para revisar visualmente el contenido
print(df.head())   # primeras 5 filas
print(df.tail())   # últimas 5 filas

# 5. Resumen estadístico rápido de las columnas numéricas
#    (cuenta, promedio, mínimo, máximo, percentiles)
print(df.describe())

# 6. Información general: tipos de dato + valores nulos por columna
df.info()

# 7. ¿Cuántos valores nulos hay por columna?
print(df.isnull().sum())
```
