```python
#Filtrar las transacciones de un país distinto a United Kingdom con Quantity > 100
import pandas as pd

df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")

# Dos condiciones combinadas con & (AND), cada una entre paréntesis:
# 1. Country distinto de "United Kingdom" → se usa != (no igual a)
# 2. Quantity mayor a 100
filtro = (df["Country"] != "United Kingdom") & (df["Quantity"] > 100)

df_filtrado = df.loc[filtro]

# Vista rápida del resultado, con las columnas más relevantes para revisar
print(df_filtrado[["Invoice", "Description", "Quantity", "Country"]].head(10))

# Cuántas filas cumplen la condición, y qué países aparecen
print("\nFilas que cumplen el filtro:", df_filtrado.shape[0])
print("\nPaíses presentes en el resultado:")
print(df_filtrado["Country"].value_counts())
```

```python
#Cuantas fulas tiene el dataset donde Quantity es negativo
import pandas as pd

df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")

# Construimos la condición booleana: True donde Quantity es negativo
condicion_negativos = df["Quantity"] < 0

# .sum() sobre una Series booleana cuenta los True (True se trata como 1, False como 0)
cantidad_negativos = condicion_negativos.sum()

print("Filas con Quantity negativo:", cantidad_negativos)

# Para contextualizar la cifra, la comparamos contra el total de filas
total_filas = df.shape[0]
porcentaje = (cantidad_negativos / total_filas) * 100

print(f"Total de filas: {total_filas}")
print(f"Porcentaje del dataset: {porcentaje:.2f}%")
```

```python
import pandas as pd

df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")

# Aislamos las filas con Quantity negativo
df_negativos = df.loc[df["Quantity"] < 0]

# De esas, nos quedamos SOLO con las que NO tienen Invoice empezando en "C"
es_nota_credito = df_negativos["Invoice"].astype(str).str.startswith("C")
df_negativos_sin_explicar = df_negativos.loc[~es_nota_credito]
# El símbolo ~ invierte la condición booleana: "donde NO es nota de crédito"

print("Filas negativas sin nota de crédito:", df_negativos_sin_explicar.shape[0])

# Hipótesis 1: ¿son ajustes de inventario con StockCode especial (no un producto real)?
# En este dataset son comunes códigos como 'D' (Discount), 'M' (Manual),
# 'POST' (Postage), 'BANK CHARGES', 'AMAZONFEE', que no son productos vendidos.
print("\nStockCode más frecuentes en estas filas:")
print(df_negativos_sin_explicar["StockCode"].value_counts().head(10))

# Hipótesis 2: ¿tienen Price en 0? (frecuente en ajustes/correcciones, no en ventas reales)
print("\n¿Cuántas tienen Price = 0?:", (df_negativos_sin_explicar["Price"] == 0).sum())

# Hipótesis 3: ¿tienen CustomerID nulo? (las devoluciones de cliente real casi
# siempre tienen CustomerID; los ajustes internos de inventario, no)
print("¿Cuántas tienen CustomerID nulo?:", df_negativos_sin_explicar["Customer ID"].isnull().sum())

# Vista directa de una muestra para inspección manual
print("\nMuestra de 10 filas sin explicar:")
print(df_negativos_sin_explicar[["Invoice", "StockCode", "Description", "Quantity", "Price"]].head(10))
```

```python
import pandas as pd

df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")

df_negativos = df.loc[df["Quantity"] < 0]
es_nota_credito = df_negativos["Invoice"].astype(str).str.startswith("C")
df_sin_explicar = df_negativos.loc[~es_nota_credito]

# ¿El Invoice de estas filas es un número "normal" (como cualquier venta),
# o tiene algún prefijo especial que se nos pasó por alto?
print("Primeros caracteres de Invoice en estas filas:")
print(df_sin_explicar["Invoice"].astype(str).str[0].value_counts())

# ¿Todas las filas tienen Description vacía, o solo la muestra que vimos?
print("\n¿Cuántas tienen Description nula?:", df_sin_explicar["Description"].isnull().sum())
print("Total de filas sin explicar:", df_sin_explicar.shape[0])

# Si TODAS coinciden en Price=0, CustomerID nulo y Description nula,
# es evidencia fuerte de que es un solo patrón consistente (ajuste de inventario)
# y no varias causas mezcladas
coinciden_las_tres = (
    (df_sin_explicar["Price"] == 0) &
    (df_sin_explicar["Customer ID"].isnull()) &
    (df_sin_explicar["Description"].isnull())
)
print("\nFilas donde coinciden las 3 señales a la vez:", coinciden_las_tres.sum(), "de", df_sin_explicar.shape[0])
```

```python
#diferencia entre df.loc[3] y df.iloc[3] una vez el dataframe fue reordenado
import pandas as pd

df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")

# Reordenamos por Quantity, de mayor a menor
df_ordenado = df.sort_values(by="Quantity", ascending=False)

# --- iloc[3]: SIEMPRE la fila en la posición física 4 (cuenta desde 0) ---

print("iloc[3]:")
print(df_ordenado.iloc[3])

# --- loc[3]: la fila cuya ETIQUETA es 3 ---
# Esa etiqueta fue asignada por el índice ORIGINAL, antes de ordenar
# (era la fila #4 del archivo, en su posición original, contando desde 0).
# sort_values() reordena las filas pero conserva sus etiquetas originales,
# así que .loc[3] va a "perseguir" esa etiqueta a donde sea que haya quedado
# después del reordenamiento.
print("\nloc[3]:")
print(df_ordenado.loc[3])

# Comprobación explícita: ¿son la misma fila o no?
son_la_misma = df_ordenado.iloc[3].equals(df_ordenado.loc[3])
print(f"\n¿iloc[3] y loc[3] devuelven la misma fila?: {son_la_misma}")
```

```python
import pandas as pd

df = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")


print("Tipo de dato ANTES:", df["Country"].dtype)

# memory_usage(deep=True) es necesario para columnas de texto (object):
# sin deep=True, Pandas solo cuenta el tamaño de los punteros a los strings,
# no el tamaño real de cada string en memoria — el número saldría engañosamente bajo.
memoria_antes = df.memory_usage(deep=True)
print("\nUso de memoria ANTES (por columna, en bytes):")
print(memoria_antes)

memoria_country_antes = memoria_antes["Country"]
print(f"\nSolo la columna Country, ANTES: {memoria_country_antes:,} bytes")

# --- Conversión a category ---
# 'category' guarda cada valor único UNA sola vez en una tabla interna,
# y cada fila solo almacena un número pequeño que apunta a esa tabla.
# Tiene sentido aquí porque Country tiene pocos valores únicos
# repetidos muchísimas veces (típico caso ideal para category).
print("\nValores únicos en Country:", df["Country"].nunique())
print("Total de filas:", df.shape[0])

df["Country"] = df["Country"].astype("category")

# --- DESPUÉS: Country ya como category ---
print("\nTipo de dato DESPUÉS:", df["Country"].dtype)

memoria_despues = df.memory_usage(deep=True)
memoria_country_despues = memoria_despues["Country"]
print(f"Solo la columna Country, DESPUÉS: {memoria_country_despues:,} bytes")

# --- Comparación directa ---
ahorro_bytes = memoria_country_antes - memoria_country_despues
ahorro_porcentaje = (ahorro_bytes / memoria_country_antes) * 100

print(f"\nAhorro de memoria en Country: {ahorro_bytes:,} bytes ({ahorro_porcentaje:.1f}%)")
```
